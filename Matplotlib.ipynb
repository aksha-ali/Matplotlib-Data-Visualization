```python
import pandas as pd
import matplotlib.pyplot as plt

# print(dir(plt))
```

```python
df = pd.DataFrame({
    'Student': ['Raj', 'Raman', 'Pooja', 'Ankit', 'Sonia'],
    'Marks': [90, 60, 70, 80, 85]
})
```

```python
#line plot
plt.plot(df['Student'],df['Marks'],marker='o',markerfacecolor='r',markeredgecolor='r',markersize = 8
        )
```

```python
#line plot
plt.plot(df['Student'],df['Marks'],color ='r',linewidth ='2',linestyle ='--' )
```

```python
overs = [1,2,3,4,5,6]
runs = [10,12,15,17,20,30]
plt.plot(overs,runs,color='y',ls='--',marker='o',mfc='b',ms = 8,mec = 'b')
```

```python
overs = [1,2,3,4,5,6]
runs = [10,12,15,17,20,30]
plt.plot(overs,runs,"or--",mfc='b',ms = 8,mec = 'b')
plt.xlabel('overs')
plt.ylabel('runs')
plt.title('Scoreboard')
plt.show()
```

```python

df = pd.read_excel(r"C:\Users\ASUS\OneDrive\Desktop\dataset_df.xlsx")
df
```

```python
df = df.drop_duplicates()
df.shape
```

```python
df.isnull().sum()
```

```python
df = df.copy()

df.loc[:, 'Unit Price'] = df['Unit Price'].fillna(df['Unit Price'].mean())
df.loc[:, 'Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())
```

```python
df[['Unit Price', 'Quantity']].isna().sum()
```

```python
df['sales']=df['Unit Price']*df['Quantity']
df
```

```python
df.columns
```

```python
df.columns = df.columns.str.strip()
```

```python
print(df.columns.tolist())
```

```python
salesperson = df[['Salesperson', 'sales']]

all_persons = salesperson.groupby('Salesperson')['sales'].sum().reset_index()

all_persons
```

```python
plt.pie(
    all_persons['sales'],
    labels=all_persons['Salesperson'],
    autopct="%0.2f%%"
)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(all_persons['Salesperson'], all_persons['sales'], "or--")
plt.show()
```

```python
plt.figure(figsize=(14,8))
plt.bar(all_persons['Salesperson'], all_persons['sales'])
plt.show()
```

```python
salesperson = df[['Salesperson', 'sales']]

all_persons = salesperson.groupby('Salesperson')['sales'].sum().reset_index()

top_five_persons = all_persons.sort_values(
    by='sales',
    ascending=False
).head(5)

top_five_persons
```

```python
plt.figure(figsize=(6,5))
plt.bar(top_five_persons['Salesperson'], top_five_persons['sales'])
plt.xticks(rotation=45)
plt.show()
```
